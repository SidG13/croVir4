```python
#placed_chromosomes_vir = ["scaffold-ma1", "scaffold-ma2", "scaffold-ma3", "scaffold-ma4", "scaffold-ma5", 
#                             "scaffold-ma6", "scaffold-ma7", "scaffold-Z", "scaffold-mi1", "scaffold-mi2", 
#                             "scaffold-mi3", "scaffold-mi4", "scaffold-mi5", "scaffold-mi6", "scaffold-mi7",
#                             "scaffold-mi8", "scaffold-mi9", "scaffold-mi10"]

placed_chromosomes_adam = ["CM077918.1", "CM077919.1", "CM077920.1", "CM077921.1", "CM077922.1", 
                           "CM077923.1", "CM077924.1", "CM077925.1", "CM077926.1", "CM077927.1", 
                           "CM077928.1", "CM077929.1", "CM077930.1", "CM077931.1", "CM077932.1", 
                           "CM077933.1", "CM077934.1", "CM077935.1", "CM077936.1"]

placed_records = []
unplaced_records = []

# Remove "RagTag" for now and separate the placed and unplaced scaffolds. 
with open(original_file) as handle:
    for record in SeqIO.parse(handle, "fasta"):
        # Remove both "_RagTag" and "_RagTag_RagTag"
        record.id = re.sub(r'_RagTag(_RagTag)?', '', record.id)
        record.description = re.sub(r'_RagTag(_RagTag)?', '', record.description)
        record.name = re.sub(r'_RagTag(_RagTag)?', '', record.name)

        print(record.id)

        # Categorize the records
        if record.id not in placed_chromosomes_adam:
            unplaced_records.append(record)
        else:
            placed_records.append(record)
```

```python
placed_records_sorted = sorted(placed_records, key=lambda record: len(record.seq), reverse=True)

for record in placed_records_sorted:
    print(record.id, ':', len(record.seq))
```

```python
# Sort unplaced scaffolds by size
sorted_unplaced_records = sorted(unplaced_records, key=lambda x: len(x.seq), reverse = True)

sorted_unplaced_records_renamed = []
un_counter = 1
for record in sorted_unplaced_records:
    record.id = f"scaffoldun{un_counter}"
    record.description = f"scaffoldun{un_counter}"
    un_counter +=1
    print(record.id, ':', len(record.seq))
    sorted_unplaced_records_renamed.append(record)
```

```python
# Merge the two sets, and write to a file
concatenated_seqs = sorted_placed_records_by_table + sorted_unplaced_records_renamed

with open(corrected_file, "w") as output_handle:
    for record in concatenated_seqs:
        record.description = ""  # Remove any existing description
        record.name = ""  # Remove any existing name
        SeqIO.write(record, output_handle, "fasta")
output_handle.close()
```
